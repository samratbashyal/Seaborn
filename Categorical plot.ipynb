```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

```python
# Categorical scatter plot
```

```python
# strip plot
# axes level function
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
# using catplot
# figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
# jitter
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=0.2,hue='sex')
```

```python
# swarmplot
sns.swarmplot(data=tips, x='day',y='total_bill')
# or
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
# Categorical distribution plot
```

```python
# Box plot --> Axes level function
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
# Using catplot --> Figure level function
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
# hue
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

```python
# Violin plot = boxplot + kdeplot
```

```python
# violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

```python
# hue

sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

```python
# Categorical Estimate plot
```

```python
# Bar plot
sns.barplot(data=tips, x='sex', y='total_bill',hue='smoker',estimator=np.mean)
# ci=none foe removing error bars
```

```python
# point plot
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker',estimator=np.mean)
```

```python
# countplot
sns.countplot(data=tips,x='sex',hue='day')
```

```python
# faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='box',row='time')
```
